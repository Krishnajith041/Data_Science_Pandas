```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = {
    "Name":["Anu","Rahul","Meera","Arun","Bob"],
    "Age": [18,19,20,21,22],
    "Marks": [85,72,60,90,65],
    "Attendance": [95,80,75,86,80],
    "City": ["Kochi","Delhi","Mumbai","Kochi","Chennai"],

}
sns.barplot(x="Attendance",y="Marks",data=data)
plt.show()
```

```python
data={
   "product" : ["laptop","phone","tablet","laptop","phone","tablet"],
   "sales" : [80,50,30,58,50,90],
   "price" : [5000,300,4000,2000,6000,8000],
   "month" : ["jan","feb","mar","apr","may","jun"]
}
sns.barplot(x="product",y="sales",data=data)

sns.countplot(x="price",data=data)


plt.show()
```

```python
sns.histplot(x="product",data=data)
```

```python
sns.boxplot(x=data["sales"])
```

```python
sns.scatterplot(x="product",y="sales",data=data)
```

```python
sns.lineplot(x="product",y="sales",data=data)
```

```python
import numpy as np
data = np.random.randn(1000)
plt.figure(figsize=(10,6))
sns.histplot(data,bins=30,kde=True,color='skyblue',edgecolor='black')
plt.title("distribution of random data using seaborn")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(axis='y',alpha=0.75)
plt.show()
```

```python
product = ["laptop","phone","tablet"]
sales = [80,50,30]

plt.pie(sales,labels=product,autopct='%4.2f%%')
plt.title("sales distribution")
plt.legend()
plt.show()
```

```python
x=[1,2,3,4,5]
y=[10,5,8,4,2]
plt.plot(x,y)
plt.fill_between(x,y)
plt.show()
```

```python
mark = [42,50,52,55,60,62,65,90]
plt.boxplot(mark)
plt.show()
```

```python
sales = [
    [10,20,30,40,50],
    [15,25,35,45,55],
    [20,30,40,50,60]
]
sns.heatmap(sales,annot=True,cmap="Reds")
plt.show()
```

```python
#1

product = ["laptop","phone","tablet","watch"]
sales = [50,80,40,70]

plt.bar(product,sales)
plt.xlabel("product")
plt.ylabel("sales")
plt.title("sales distribution")
plt.show()
```

```python
product = ["laptop","phone","tablet","watch"]
sales = [50,80,40,70]
plt.pie(sales,labels=product,autopct='%1.1f%%')
plt.title("sales distribution")
plt.legend()
plt.show()
```

```python
#2
days = ["mon","tue","wed","thu","fri"]
visitors = [120,120,180,160,120]

plt.plot(days,visitors)
plt.xlabel("days")
plt.ylabel("visitors")
plt.title("visitors distribution")
plt.show()
```

```python
#3
categories = ["food","travel","rent","shopping"]
expenses = [5000,3000,8000,4000]

plt.pie(expenses,labels=categories,autopct='%4.2f%%')
plt.title("expenses distribution")

plt.show()
```

```python
advertising = [10,20,30,40,50]
sales = [25,35,45,55,70]

plt.scatter(advertising,sales)
plt.xlabel("advertising")
plt.ylabel("sales")
plt.title("advertising vs sales")
plt.show()
```
