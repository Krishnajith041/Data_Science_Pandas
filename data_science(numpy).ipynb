importing numpy

```python
import numpy as np
```

```python
arr=np.array([1,2,3])
print(arr)
```

```python
arr1=np.array([1,2,3])
arr2=np.array([4,5,6])
print(arr1+arr2)
print(arr1-arr2)
print(arr1*arr2)
print(arr1/arr2)
print(arr1**arr2)
```

2.COMPARISON OPERATORS

```python
arr1=np.array([1,2,3])
arr2=np.array([4,5,6])
print(arr1>arr2)
print(arr1==arr2)
print(arr1!=arr2)
```

ARRAY SHAPE OPERATION

```python
arr1=np.array([1,2,3])                         #1d array
arr2=np.array([[1,2,3],[5,4,6]])               #2d array
arr3=np.array([[[1,2,3],[5,4,6],[5,4,6]]])     #3d array
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)
```

```python
arr5=np.array([[[[[1,2,3],[5,4,2],[5,4,6],[3,5,7],[4,6,9]]]]])   #5d array
print(arr5.shape)
```
