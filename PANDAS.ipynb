```python
import pandas as pd
```

```python
data=[20,40,60,100]
s=pd.Series(data,index=['a','b','c','d'])
print(s)
```

```python
data={"NAME":"krishnajith","AGE":18}
s=pd.Series(data)
print(s)
```

```python
data=[20,30,40,50,60]
s=pd.Series(data)
print(s)
print(s.loc[1])
```

```python
data=[20,30,40,50,60]
s=pd.Series(data,index=['a','b','c','d','e'])
print(s)
print(s.iloc[3])
```

```python
data=[20,30,40,50,60]
s=pd.Series(data,index=['a','b','c','d','e'])                #slice
print(s.loc["a":"d"])
print(s.iloc[:3])
print(s.loc[s<40])
```

```python
sales=pd.Series([900,1000,3000,4000],index=["Monday","Tuesday","Wednesday","Thursday"])
print(sales)
print(sales.max())
print(sales.min())
print(sales.sum())
print(sales[sales<1000])
```

```python
data = [20,30,30,30,20,40,50,50,60,40]
ar=pd.Series(data)
print(ar)
print(ar.unique())
print(ar.nunique())   #to get the count
print(ar.value_counts())  #to get the unique values
```

```python
data = [20,30,None,30,20,40,None,50,60,40,None,None]
ar=pd.Series(data)
print(ar)
print(ar.isnull())
```

```python
data = [20,30,None,30,20,40,None,50,60,40,None,None]
ar=pd.Series(data)
print(ar)
print(ar.isna())
```

```python
data = [20,30,None,30,20,40,None,50,60,40,None,None]
ar=pd.Series(data)
print(ar)
print(ar.notna())
```

```python
data = [20,30,None,30,20,40,None,50,60,40,None,None]
ar=pd.Series(data)
print(ar)
print(ar.dropna())
```
