# ***Data Frame***

```python
import numpy as np
data = {
    "Name": ["Anu","Rahul","Meera","Arun","Bob"],
    "Age": [18,19,20,21,22],
    "Marks": [85,72,np.nan,90,65],
    "Attendance": [95,80,75,np.nan,80],
    "City": ["Kochi","Delhi","Mumbai","Kochi","Chennai"]
}

df = pd.DataFrame(data)
print(df)
```

```python
import numpy as np
data = {
    "Name":["Anu","Rahul","Meera","Arun","Bob"],
    "Age": [18,19,20,21,22],
    "Marks": [85,72,np.nan,90,65],
    "Attendance": [95,80,75,np.nan,80],
    "City": ["Kochi","Delhi","Mumbai","Kochi","Chennai"],

}

df = pd.DataFrame(data)
print(df[["Name","Marks"]])
```
