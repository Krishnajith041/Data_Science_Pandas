```python
data=[20,30,40,50,60]
s=pd.Series(data,index=['a','b','c','d','e'])
print(s)
print(s.iloc[3])
```

```python
data=[20,30,40,50,60]
s=pd.Series(data,index=['a','b','c','d','e'])                #slice
print(s.loc["a":"d"])
print(s.iloc[:3])
```
